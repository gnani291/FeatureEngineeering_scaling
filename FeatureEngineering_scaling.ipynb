```python
import pandas as pd
data={
    "name":["ram","sam","tom"],
    "city":["mumabi","hyd","delhi"]

}
df=pd.DataFrame(data)
print(df)
```

```python
# Label encoding
import pandas as pd
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df["city_encoded"]=le.fit_transform(df["city"])
print(df)
```

```python
#one hot encoding
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder()
encoded=encoder.fit_transform(df[["city"]])
print(encoded.toarray())
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False)

encoded = encoder.fit_transform(df[["city"]])

encoded_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(["city"])
)

df = pd.concat([df, encoded_df], axis=1)

print(df)
```

```python
import pandas as pd

df = pd.DataFrame({
    "City": ["Mumbai", "Delhi", "Hyderabad", "Mumbai", "Delhi"],
    "Education": ["Bachelor", "Master", "PhD", "Bachelor", "Master"]
})

print(df)
```

```python
#ordinal encoding
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
Oe=OrdinalEncoder()
df["Education_order"]=Oe.fit_transform(df[["Education"]])
print(df)
```

```python
#feature encoding
freq = df["City"].value_counts()

df["City_frequency"] = df["City"].map(freq)

print(df)
```

```python
#Target encoding
from sklearn.preprocessing import TargetEncoder
y=[1,0,1,1,0]
encoder=TargetEncoder(cv=2)
encoded=encoder.fit_transform(df[["City"]],y)
print(encoded)
```

```python
import pandas as pd
data={
    "name":["ram","sam","tom","raju"],
    "salary":[100000,200000,300000,400000]

}
df=pd.DataFrame(data)
print(df)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [12, 18, 25, 35, 45, 60, 75]
})

print(df)
```

```python
data={
    "name":["ram","sam","tom","raju"],
    "salary":[100000,200000,300000,1000000]

}
df=pd.DataFrame(data)
print(df)
```

```python
#handling outliers using IQR method
import pandas as pd
q1=df["salary"].quantile(0.25)
q3=df["salary"].quantile(0.75)
IQR=q3-q1
lower=q1-1.5*IQR
upper=q3+1.5* IQR
outliers=df[(df["salary"]<lower)|(df["salary"]>upper)]
print(outliers)
```

```python
df["salary"]=df["salary"].clip(lower,upper)
print(df)
```

```python
import numpy as np
df["salary"]=np.log(df["salary"])
print(df)
```

```python
import pandas as pd
data={
    "name":["ram","sam","tom"],
    "income":[1000,2000,400000]
}
df=pd.DataFrame(data)
print(df)
```
